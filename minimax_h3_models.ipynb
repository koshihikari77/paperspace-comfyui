# MiniMax H3 モデル取得

研究用 ComfyUI (`/storage/h3-research`) に使う重みを取得します。通常の Wan 環境 (`/storage/ComfyUI`, `/app/models`) は変更しません。既定の `fused-core` は無音 I2V/Ref2VA に必要な3ファイルです。音声出力や他のモデル系統を使うときだけ追加グループを選びます。

取得済みのファイルはサイズを確認して再利用し、新規取得は SHA-256 で照合します。取得済みファイルも完全照合したい場合は `VERIFY_EXISTING = True` にしてください。


```python
from pathlib import Path
import subprocess
import sys

REPO_ROOT = Path("/notebooks")
MODEL_ROOT = Path("/storage/h3-research/models")
MODEL_GROUPS = ["fused-core"]  # 追加: "audio", "official-fl2va", "official-ref2va", "fasth3"
VERIFY_EXISTING = False

def download_command(dry_run=False):
    command = [sys.executable, str(REPO_ROOT / "scripts/download_minimax_h3.py"), "--model-root", str(MODEL_ROOT)]
    for group in MODEL_GROUPS:
        command.extend(["--group", group])
    if VERIFY_EXISTING:
        command.append("--verify-existing")
    if dry_run:
        command.append("--dry-run")
    return command

print("Selected groups:", ", ".join(MODEL_GROUPS))
print("Model root:", MODEL_ROOT)
```

```python
subprocess.run(download_command(dry_run=True), check=True)
```

## ダウンロード

次のセルを実行すると、不足分だけ取得します。数十GBのグループは長時間かかります。中断時は `.partial` から再開します。既存ファイルがサイズやSHA-256と一致しなければ上書きせず停止します。


```python
subprocess.run(download_command(), check=True)
```

取得先は研究用 ComfyUI の `model-paths.yaml` で参照されています。`fused-core` だけで無音動画を作れます。音声付きワークフローには `audio` を追加してください。公式ベースや FastH3 のグループは専用の生成checkpointを追加します。対応関係は `/storage/h3-research/H3-MODELS-I2V-REF2VA-2026-09-24.md` を参照してください。
